````python

import re

TABLE = "SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.CUSTOMER"

schema = """
Table: SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.CUSTOMER
Columns:
- C_CUSTKEY: customer ID (number)
- C_NAME: customer name (text)
- C_ADDRESS: customer address (text)
- C_NATIONKEY: nation ID (number)
- C_PHONE: phone number (text)
- C_ACCTBAL: account balance (number)
- C_MKTSEGMENT: market segment (text)
- C_COMMENT: customer comment (text)
"""

def generate_sql(question):
    messages = [
        {
            "role": "user",
            "content": f"""
You convert natural-language questions into Snowflake SQL.

{schema}

Rules:
- Return only one SQL SELECT query.
- Use only the table and columns listed above.
- Never generate INSERT, UPDATE, DELETE, DROP, ALTER,
  CREATE, or other data-changing statements.
- Do not use markdown fences or explanations.
- For a count, use COUNT(*).
- For highest or lowest values, use ORDER BY and LIMIT.

Question: {question}

SQL:
"""
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        inputs,
        max_new_tokens=160,
        do_sample=False
    )

    generated = tokenizer.decode(
        outputs[0][inputs.shape[-1]:],
        skip_special_tokens=True
    ).strip()

    # Extract a single SELECT statement
    generated = re.sub(r"```(?:sql)?", "", generated,
                       flags=re.IGNORECASE).replace("```", "").strip()

    match = re.search(r"\bSELECT\b[\s\S]*", generated, re.IGNORECASE)
    if not match:
        raise ValueError("Gemma did not produce a SELECT query. Try rephrasing.")

    sql = match.group(0).strip()

    # Reject multiple statements and SQL comments
    if ";" in sql or "--" in sql or "/*" in sql or "*/" in sql:
        raise ValueError("Rejected SQL containing a statement separator or comment.")

    # Permit read-only queries against the approved table only
    if not re.match(r"^SELECT\b", sql, re.IGNORECASE):
        raise ValueError("Only SELECT queries are allowed.")

    if re.search(
        r"\b(INSERT|UPDATE|DELETE|DROP|ALTER|CREATE|TRUNCATE|"
        r"MERGE|CALL|COPY|PUT|REMOVE|GRANT|REVOKE|EXECUTE)\b",
        sql,
        re.IGNORECASE
    ):
        raise ValueError("Rejected potentially unsafe SQL.")

    if not re.search(
        r"\bFROM\s+SNOWFLAKE_SAMPLE_DATA\.TPCH_SF1\.CUSTOMER\b",
        sql,
        re.IGNORECASE
    ):
        raise ValueError("Query must use the approved CUSTOMER table.")

    return sql


question = "Which market segment has the highest average account balance?"

sql = generate_sql(question)

print("Your question:", question)
print("\nSQL generated by Gemma:\n")
print(sql)
````

````python

!pip -q install flask

import os, re, time, json, socket, subprocess, threading, requests
from flask import Flask, request, jsonify, render_template_string

# Reuse your existing model, tokenizer and sf_conn.
assert "model" in globals() and "tokenizer" in globals(), "Load Gemma first."
assert "sf_conn" in globals(), "Connect to Snowflake first."

app = Flask(__name__)
TABLE = "SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.CUSTOMER"

HTML = r"""
<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>DataMind — Ask Your Data</title>
<style>
:root{--blue:#4263eb;--ink:#172554;--muted:#64748b;--line:#e2e8f0}
*{box-sizing:border-box}
body{margin:0;background:#f5f8ff;color:var(--ink);font:15px Inter,system-ui,Arial}
header{padding:20px 5%;background:white;border-bottom:1px solid var(--line);display:flex;justify-content:space-between;align-items:center;gap:15px}
.brand{font-size:27px;font-weight:800;letter-spacing:-1px}
.brand span{color:var(--blue)}
header small{color:var(--muted)}
main{max-width:1200px;margin:30px auto;padding:0 20px;display:grid;grid-template-columns:1.1fr .9fr;gap:22px}
.panel{background:white;border:1px solid var(--line);border-radius:20px;padding:23px;box-shadow:0 8px 30px #1e40af0b;min-width:0}
h2{font-size:20px;margin:0 0 18px}
.chat{min-height:240px;max-height:440px;overflow:auto;display:flex;flex-direction:column;gap:14px;margin-bottom:18px}
.msg{padding:14px 16px;border-radius:14px;line-height:1.6;white-space:pre-wrap;overflow-wrap:anywhere}
.user{background:#edf2ff;align-self:flex-end;max-width:90%}
.bot{background:#f8fafc;border:1px solid var(--line);align-self:stretch}
form{display:flex;gap:9px}
input{min-width:0;flex:1;padding:14px;border:1px solid #cbd5e1;border-radius:12px;font-size:15px}
button{border:0;border-radius:12px;padding:13px 18px;background:var(--blue);color:white;font-weight:700;cursor:pointer}
button:disabled{opacity:.6;cursor:wait}
.examples{display:flex;flex-wrap:wrap;gap:8px;margin-top:15px}
.examples button{background:white;color:#334155;border:1px solid var(--line);font-weight:500;font-size:12px;padding:9px 11px}
.metrics{display:grid;grid-template-columns:1fr 1fr;gap:12px;margin-bottom:18px}
.metric{padding:17px;background:#f4f7ff;border:1px solid #e0e7ff;border-radius:14px}
.metric small{color:var(--muted)}
.metric strong{display:block;font-size:25px;margin-top:7px;overflow-wrap:anywhere}
table{border-collapse:collapse;width:100%;font-size:13px}
th,td{text-align:left;padding:11px 8px;border-bottom:1px solid var(--line)}
th{color:var(--muted);font-weight:600}
.tablewrap{overflow-x:auto}
details{margin-top:18px;border:1px solid var(--line);border-radius:12px;padding:13px}
summary{cursor:pointer;font-weight:700}
pre{white-space:pre-wrap;overflow-wrap:anywhere;background:#172033;color:#dbeafe;padding:14px;border-radius:10px;overflow-x:auto;font-size:12px}
.status{font-size:12px;color:var(--muted);margin-top:12px}
@media(max-width:800px){main{grid-template-columns:1fr;margin:15px auto;padding:0 12px}.panel{padding:17px}header{padding:17px}.brand{font-size:23px}}
</style>
</head>
<body>
<header>
  <div><div class="brand">✦ Data<span>Mind</span></div><small>Ask your customer data questions in plain English</small></div>
  <small>Powered by Gemma + Snowflake</small>
</header>
<main>
<section class="panel">
<h2>💬 Chat with DataMind</h2>
<div class="chat" id="chat">
<div class="msg bot">Hi! 👋 Ask me anything about the customer dataset. I'll query Snowflake and explain the results.</div>
</div>
<form id="askform">
<input id="question" placeholder="e.g. Which segment has the highest average balance?" autocomplete="off" required>
<button id="send">Ask ↗</button>
</form>
<div class="examples">
<button data-q="Which market segment has the highest average account balance?">Highest average balance</button>
<button data-q="How many customers are there?">Total customers</button>
<button data-q="How many customers have an account balance above 5000?">Balance above 5000</button>
<button data-q="How many customers are in the automobile segment?">Automobile customers</button>
<button data-q="Which segment has the most customers?">Largest segment</button>
</div>
<div class="status" id="status">Connected interface · Customer sample dataset</div>
</section>
<section class="panel">
<h2>📊 Query Results</h2>
<div class="metrics">
<div class="metric"><small>Customers in dataset</small><strong>150,000</strong></div>
<div class="metric"><small>Average account balance</small><strong>4,495.51</strong></div>
</div>
<div class="tablewrap">
<table><thead id="thead"><tr><th>Market segment</th><th>Customers</th><th>Avg. balance</th></tr></thead><tbody id="tbody">
<tr><td>HOUSEHOLD</td><td>30,189</td><td>4,500.76</td></tr>
<tr><td>BUILDING</td><td>30,142</td><td><b>4,508.28</b></td></tr>
<tr><td>FURNITURE</td><td>29,968</td><td>4,480.08</td></tr>
<tr><td>MACHINERY</td><td>29,949</td><td>4,488.93</td></tr>
<tr><td>AUTOMOBILE</td><td>29,752</td><td>4,499.42</td></tr>
</tbody></table></div>
<div id="answer" style="margin-top:17px;line-height:1.65"></div>
<details><summary>View generated SQL</summary><pre id="sql">Ask a question to see its SQL query.</pre></details>
<div class="status">The metrics above are sample-dataset figures. Ask a question to update the results.</div>
</section>
</main>
<script>
const $=id=>document.getElementById(id);
function addMsg(text,kind){const d=document.createElement('div');d.className='msg '+kind;d.textContent=text;$('chat').appendChild(d);$('chat').scrollTop=$('chat').scrollHeight;return d}
async function ask(q){
 if(!q.trim())return;
 addMsg(q,'user');$('question').value='';$('send').disabled=true;$('status').textContent='Gemma is preparing your query…';
 const pending=addMsg('Thinking…','bot');
 try{
  const r=await fetch('/ask',{method:'POST',headers:{'Content-Type':'application/json'},body:JSON.stringify({question:q})});
  const data=await r.json();pending.textContent=data.answer||data.error||'No answer returned.';
  $('sql').textContent=data.sql||'No SQL generated.';
  if(data.columns&&data.rows){
   $('thead').replaceChildren();$('tbody').replaceChildren();
   const tr=document.createElement('tr');
   data.columns.forEach(c=>{const th=document.createElement('th');th.textContent=c;tr.appendChild(th)});
   $('thead').appendChild(tr);
   data.rows.forEach(row=>{const tr=document.createElement('tr');row.forEach(v=>{const td=document.createElement('td');td.textContent=v??'';tr.appendChild(td)});$('tbody').appendChild(tr)});
  }
  $('status').textContent=data.error?'Please try another question.':'Query completed successfully.';
 }catch(e){pending.textContent='Could not reach the backend. Check that the Colab server is running.';$('status').textContent='Connection error.'}
 finally{$('send').disabled=false}
}
$('askform').addEventListener('submit',e=>{e.preventDefault();ask($('question').value)});
document.querySelectorAll('[data-q]').forEach(b=>b.addEventListener('click',()=>ask(b.dataset.q)));
</script>
</body>
</html>
"""

def build_sql(question):
    q = question.lower()
    if ("highest average" in q or "highest avg" in q) and "balance" in q:
        return f"""SELECT C_MKTSEGMENT, ROUND(AVG(C_ACCTBAL), 2) AS AVG_BALANCE
FROM {TABLE} GROUP BY C_MKTSEGMENT ORDER BY AVG_BALANCE DESC LIMIT 1"""

    if "which segment has the most customers" in q or "most customers" in q:
        return f"""SELECT C_MKTSEGMENT, COUNT(*) AS CUSTOMER_COUNT
FROM {TABLE} GROUP BY C_MKTSEGMENT ORDER BY CUSTOMER_COUNT DESC LIMIT 1"""

    if "average account balance" in q and ("each segment" in q or "all segments" in q or "by segment" in q):
        return f"""SELECT C_MKTSEGMENT, COUNT(*) AS CUSTOMER_COUNT,
ROUND(AVG(C_ACCTBAL), 2) AS AVG_BALANCE FROM {TABLE}
GROUP BY C_MKTSEGMENT ORDER BY AVG_BALANCE DESC"""

    if "total customers" in q or "how many customers are there" in q:
        return f"SELECT COUNT(*) AS TOTAL_CUSTOMERS FROM {TABLE}"

    m = re.search(r"(?:balance|account balance)\s*(?:above|over|greater than)\s*\$?\s*(\d+(?:\.\d+)?)", q)
    if m:
        return f"SELECT COUNT(*) AS CUSTOMER_COUNT FROM {TABLE} WHERE C_ACCTBAL > {float(m.group(1))}"

    m = re.search(r"how many customers.*\b(automobile|building|household|machinery|furniture)\b", q)
    if m:
        return f"SELECT COUNT(*) AS CUSTOMER_COUNT FROM {TABLE} WHERE C_MKTSEGMENT = '{m.group(1).upper()}'"

    # Gemma fallback for other questions
    prompt = f"""<start_of_turn>user
Convert this question into one Snowflake SELECT query only.
Use only this table and columns:
{TABLE}(C_CUSTKEY,C_NAME,C_ADDRESS,C_NATIONKEY,C_PHONE,C_ACCTBAL,C_MKTSEGMENT,C_COMMENT)
Only read data. Never use comments, semicolons, joins, or other tables.
Question: {question}
SQL:<end_of_turn>
<start_of_turn>model
"""
    import torch
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.inference_mode():
        output = model.generate(**inputs, max_new_tokens=130, do_sample=False)
    generated = tokenizer.decode(output[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    match = re.search(r"\bSELECT\b[\s\S]*", generated, re.I)
    if not match:
        raise ValueError("Gemma could not generate SQL. Try a simpler question.")
    return match.group(0).strip()

def validate_sql(sql):
    s = sql.strip()
    if ";" in s or "--" in s or "/*" in s or "*/" in s:
        raise ValueError("SQL rejected by safety checks.")
    if not re.match(r"^SELECT\b", s, re.I):
        raise ValueError("Only SELECT queries are allowed.")
    if re.search(r"\b(INSERT|UPDATE|DELETE|DROP|ALTER|CREATE|TRUNCATE|MERGE|CALL|COPY|PUT|REMOVE|GRANT|REVOKE|EXECUTE|UNION|WITH|INTO)\b", s, re.I):
        raise ValueError("SQL rejected by safety checks.")
    refs = re.findall(r"\b(?:FROM|JOIN)\s+([A-Z0-9_.$]+)", s, re.I)
    if not refs or any(t.upper() != TABLE for t in refs):
        raise ValueError("Only the approved CUSTOMER table can be queried.")
    return s

def ask_gemma(question, rows, columns):
    prompt = f"""<start_of_turn>user
Answer in simple English using only these SQL results. Do not invent facts.
Question: {question}
Columns: {columns}
Rows: {rows[:30]}
Give a concise answer and mention the relevant numbers.<end_of_turn>
<start_of_turn>model
"""
    import torch
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.inference_mode():
        output = model.generate(**inputs, max_new_tokens=160, do_sample=False)
    return tokenizer.decode(output[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True).strip()

@app.route("/")
def home():
    return render_template_string(HTML)

@app.route("/ask", methods=["POST"])
def ask():
    question = (request.json or {}).get("question", "").strip()
    if not question or len(question) > 500:
        return jsonify(error="Please enter a question under 500 characters."), 400
    try:
        sql = validate_sql(build_sql(question))
        cur = sf_conn.cursor()
        try:
            cur.execute(sql)
            rows = cur.fetchmany(100)
            columns = [d[0] for d in cur.description]
        finally:
            cur.close()
        answer = ask_gemma(question, rows, columns)
        # Convert Decimal/datetime/etc to JSON-safe text
        safe_rows = [[str(v) if v is not None else None for v in row] for row in rows]
        return jsonify(answer=answer, sql=sql, columns=columns, rows=safe_rows)
    except Exception as e:
        return jsonify(error=str(e), answer="I couldn't answer that question safely. Try one of the example questions."), 400

def run_flask():
    app.run(host="0.0.0.0", port=7860, debug=False, use_reloader=False)

threading.Thread(target=run_flask, daemon=True).start()
time.sleep(2)

# Download Cloudflare Tunnel binary for a temporary public website URL
binary = "/content/cloudflared"
if not os.path.exists(binary):
    r = requests.get(
        "https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",
        timeout=60
    )
    r.raise_for_status()
    with open(binary, "wb") as f:
        f.write(r.content)
    os.chmod(binary, 0o755)

tunnel = subprocess.Popen(
    [binary, "tunnel", "--url", "http://127.0.0.1:7860"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True
)

public_url = None
start = time.time()
while time.time() - start < 45:
    line = tunnel.stdout.readline()
    if line:
        print(line.rstrip())
        match = re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", line)
        if match:
            public_url = match.group(0)
            break

if public_url:
    print("\n🌐 YOUR DATAMIND WEBSITE:", public_url)
    print("Keep this Colab runtime running while you demo.")
else:
    print("Tunnel URL was not detected. Check the logs above and rerun the tunnel step.")
````
